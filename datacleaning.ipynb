```python

df = pd.read_csv("data/online_retail_raw.csv", index_col=False)

# Cancellation invoices start with "C"; keep them separate from the sales analysis.
is_cancellation = (
    df["InvoiceNo"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.startswith("C", na=False)
)

cancelled_df = df.loc[is_cancellation].copy()
df = df.loc[~is_cancellation].copy()

print(f"Removed {len(cancelled_df):,} cancellation rows.")
print(f"Rows remaining for analysis: {len(df):,}")
df.head(10)
```

```python
code = df["StockCode"].astype("string").str.strip()

# Numeric-only stock codes already present in the data
numeric_codes = set(code[code.str.fullmatch(r"\d+", na=False)].dropna())

review = (
    df.loc[
        code.str.contains(r"[A-Za-z]", na=False),
        ["StockCode", "Description"]
    ]
    .assign(
        NumericPart=lambda rows: rows["StockCode"]
            .astype("string")
            .str.replace(r"\D", "", regex=True)
    )
    .value_counts()
    .rename("rows")
    .reset_index()
)

# True means the digits from this code also exist as a numeric-only code
review["HasNumericOnlyMatch"] = (
    review["NumericPart"].ne("")
    & review["NumericPart"].isin(numeric_codes)
)

# Show possible matches first
review.sort_values(
    ["HasNumericOnlyMatch", "rows"],
    ascending=[False, False]
)
```

```python
code = df["StockCode"].astype("string").str.strip()
has_letters = code.str.contains(r"[A-Za-z]", na=False)
numeric_part = code.str.replace(r"\D", "", regex=True)

# Numeric-only stock codes already present in the data
numeric_codes = set(code[code.str.fullmatch(r"\d+", na=False)].dropna())

# Include lettered codes and numeric-only codes that match their digits
lettered_numeric_parts = set(numeric_part[has_letters].dropna()) - {""}
show_rows = has_letters | (
    code.isin(lettered_numeric_parts) & code.isin(numeric_codes)
)

review_rows = df.loc[show_rows, ["StockCode", "Description"]].copy()
review_rows["StockCode"] = code.loc[show_rows]
review_rows["NumericPart"] = numeric_part.loc[show_rows]

review = (
    review_rows
    .value_counts()
    .rename("rows")
    .reset_index()
)

# True on a lettered code when its numeric-only version exists
review["HasNumericOnlyMatch"] = (
    review["StockCode"].str.contains(r"[A-Za-z]", na=False)
    & review["NumericPart"].isin(numeric_codes)
)

# Put codes with the same digits together
review = review.sort_values(
    ["NumericPart", "StockCode", "Description"],
    kind="stable",
    na_position="last"
)

review.head(30)
```

```python
nonpositive_quantity = df["Quantity"] <= 0
nonpositive_price = df["UnitPrice"] <= 0
rows_to_remove = nonpositive_quantity | nonpositive_price

print(f"Rows with zero or negative quantity: {nonpositive_quantity.sum():,}")
print(f"Rows with zero or negative price: {nonpositive_price.sum():,}")
print(f"Rows with both: {(nonpositive_quantity & nonpositive_price).sum():,}")
print(f"Unique rows to remove: {rows_to_remove.sum():,}")

display(df.loc[rows_to_remove])
```

```python
rows_before = len(df)
df = df.loc[~rows_to_remove].copy()

print(f"Rows removed: {rows_before - len(df):,}")
print(f"Rows remaining: {len(df):,}")
```

```python
# Make sure dates are real datetime values, then add revenue and year-month columns
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="raise")
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")
```

```python
last_month = df["InvoiceDate"].max().to_period("M")

monthly_revenue = (
    df.groupby("YearMonth", as_index=False)["Revenue"]
      .sum()
      .sort_values("YearMonth")
)

monthly_revenue["Month"] = monthly_revenue["YearMonth"].astype(str)

monthly_revenue.loc[
    monthly_revenue["YearMonth"] == last_month, "Month"
] += " (partial)"

monthly_revenue[["Month", "Revenue"]]
```

```python
top_30_products = (
    df.groupby(["StockCode", "Description"], as_index=False, dropna=False)["Revenue"]
      .sum()
      .sort_values("Revenue", ascending=False)
      .head(30)
)

top_30_products
```

```python
top_20_countries = (
    df.groupby("Country", as_index=False)["Revenue"]
      .sum()
      .sort_values("Revenue", ascending=False)
      .head(20)
)

top_20_countries
```

```python
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Where to save the results
output_dir = Path(
    r"C:\Users\USER\Desktop\Data Science Challenge\Week 1 (online-retail-report)\output"
)
output_dir.mkdir(parents=True, exist_ok=True)

# Format chart axes as GBP
gbp_format = FuncFormatter(lambda value, position: f"£{value:,.0f}")

# Prepare copies for Excel and label revenue in GBP
monthly_excel = monthly_revenue[["Month", "Revenue"]].copy()
monthly_excel = monthly_excel.rename(columns={"Revenue": "Revenue (GBP)"})

products_excel = top_10_products[
    ["StockCode", "Description", "Revenue"]
].copy()
products_excel = products_excel.rename(columns={"Revenue": "Revenue (GBP)"})

countries_excel = top_10_countries[["Country", "Revenue"]].copy()
countries_excel = countries_excel.rename(columns={"Revenue": "Revenue (GBP)"})

# 1. Revenue per month
fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(monthly_revenue["Month"], monthly_revenue["Revenue"], color="steelblue")
ax.yaxis.set_major_formatter(gbp_format)
ax.set_title("Revenue per Month (GBP)")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (GBP)")
plt.xticks(rotation=45, ha="right")
fig.tight_layout()
fig.savefig(output_dir / "revenue_per_month.png", dpi=300)
plt.close(fig)

# 2. Top 10 products by revenue
products_plot = top_10_products.copy()
products_plot["Product"] = (
    products_plot["StockCode"].astype("string")
    + " - "
    + products_plot["Description"].fillna("Unknown description").astype("string")
)
products_plot = products_plot.sort_values("Revenue")

fig, ax = plt.subplots(figsize=(12, 7))
ax.barh(products_plot["Product"], products_plot["Revenue"], color="seagreen")
ax.xaxis.set_major_formatter(gbp_format)
ax.set_title("Top 10 Products by Revenue (GBP)")
ax.set_xlabel("Revenue (GBP)")
ax.set_ylabel("Product")
fig.tight_layout()
fig.savefig(output_dir / "top_10_products_by_revenue.png", dpi=300)
plt.close(fig)

# 3. Top 10 countries by revenue
countries_plot = top_10_countries.sort_values("Revenue")

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(countries_plot["Country"], countries_plot["Revenue"], color="darkorange")
ax.xaxis.set_major_formatter(gbp_format)
ax.set_title("Top 10 Countries by Revenue (GBP)")
ax.set_xlabel("Revenue (GBP)")
ax.set_ylabel("Country")
fig.tight_layout()
fig.savefig(output_dir / "top_10_countries_by_revenue.png", dpi=300)
plt.close(fig)

# Save the three result tables as separate Excel sheets
with pd.ExcelWriter(output_dir / "online_retail_revenue_analysis.xlsx") as writer:
    monthly_excel.to_excel(writer, sheet_name="Monthly Revenue", index=False)
    products_excel.to_excel(writer, sheet_name="Top 10 Products", index=False)
    countries_excel.to_excel(writer, sheet_name="Top 10 Countries", index=False)

print(f"Charts and Excel workbook saved to: {output_dir}")
```
